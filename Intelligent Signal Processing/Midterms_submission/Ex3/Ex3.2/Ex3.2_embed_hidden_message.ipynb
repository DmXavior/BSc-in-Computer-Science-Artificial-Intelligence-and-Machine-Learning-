**Embedding Algorithm**

```python
import wave
import numpy as np

def embed_message(audio_file, message, output_file):
    # Read the audio file
    with wave.open(audio_file, 'rb') as wav:
        params = wav.getparams()
        frames = wav.readframes(params.nframes)
    
    # Convert frames to numpy array and create a writable copy
    audio_data = np.frombuffer(frames, dtype=np.int16).copy()  # Use .copy() to make it writable
    
    # Convert message to binary
    message_bin = ''.join(format(ord(i), '08b') for i in message)
    message_length = len(message_bin)
    
    # Check if the message can fit in the audio data
    if message_length > len(audio_data):
        raise ValueError("Message is too long to fit in the audio file.")
    
    # Embed the message into the least significant bits
    for i in range(message_length):
        audio_data[i] = (audio_data[i] & ~1) | int(message_bin[i])
    
    # Write the modified audio data to a new file
    with wave.open(output_file, 'wb') as wav_out:
        wav_out.setparams(params)
        wav_out.writeframes(audio_data.tobytes())

# Input secret message and generate embedded file
secret_message = "An eye for an eye makes the whole world blind"
embed_message('Ex3_sound5.wav', secret_message, 'Ex3_sound5_embedded.wav')
```

**Extraction Algorithm**

```python
def extract_message(audio_file, message_length):
    # Read the audio file
    with wave.open(audio_file, 'rb') as wav:
        frames = wav.readframes(wav.getnframes())
    
    # Convert frames to numpy array
    audio_data = np.frombuffer(frames, dtype=np.int16)
    
    # Extract the message from the least significant bits
    message_bin = ''
    for i in range(message_length):
        message_bin += str(audio_data[i] & 1)
    
    # Convert binary to string
    message = ''.join(chr(int(message_bin[i:i+8], 2)) for i in range(0, len(message_bin), 8))
    return message

# Extraction of secret code
extracted_message = extract_message('Ex3_sound5_embedded.wav', len(secret_message) * 8)
print("Extracted Message:", extracted_message)
```
